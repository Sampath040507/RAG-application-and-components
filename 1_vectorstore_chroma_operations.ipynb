```python
from langchain_openai import OpenAIEmbeddings
from langchain_chroma import Chroma
```

```python
from langchain_core.documents import Document #output_parsers, prompt templates, messages and docuents all are imported from core

# Create LangChain documents for IPL players

doc1 = Document(
        page_content="Virat Kohli is one of the most successful and consistent batsmen in IPL history. Known for his aggressive batting style and fitness, he has led the Royal Challengers Bangalore in multiple seasons.",
        metadata={"team": "Royal Challengers Bangalore"}
    )
doc2 = Document(
        page_content="Rohit Sharma is the most successful captain in IPL history, leading Mumbai Indians to five titles. He's known for his calm demeanor and ability to play big innings under pressure.",
        metadata={"team": "Mumbai Indians"}
    )
doc3 = Document(
        page_content="MS Dhoni, famously known as Captain Cool, has led Chennai Super Kings to multiple IPL titles. His finishing skills, wicketkeeping, and leadership are legendary.",
        metadata={"team": "Chennai Super Kings"}
    )
doc4 = Document(
        page_content="Jasprit Bumrah is considered one of the best fast bowlers in T20 cricket. Playing for Mumbai Indians, he is known for his yorkers and death-over expertise.",
        metadata={"team": "Mumbai Indians"}
    )
doc5 = Document(
        page_content="Ravindra Jadeja is a dynamic all-rounder who contributes with both bat and ball. Representing Chennai Super Kings, his quick fielding and match-winning performances make him a key player.",
        metadata={"team": "Chennai Super Kings"}
    )
```

```python
#create a list of document objects:
docs = [doc1, doc2, doc3, doc4, doc5]
```

```python
vector_store = Chroma(
    embedding_function=OpenAIEmbeddings(),
    persist_directory='my_chroma_db',
    collection_name='sample'
)
#You are creating a LangChain Chroma object that represents a Chroma vector store, and you're telling it how to create embeddings and where to persist the Chroma data.
#"Create/connect to a Chroma vector store, use OpenAI's embedding model whenever text needs to be converted into vectors, persist the database under my_chroma_db, and use a collection called sample."
```

```python
#ADD
vector_store.add_documents(docs)
#not just documents getting added to the db, at the same time, every document gets assigned a unique id so that later we can use that id to retrieve the doc
#these are default ids but if u want to specify ur own....u can!
```

```python
# VIEW documents and their info...u use "get", but u will get it with ids first
vector_store.get(include=['embeddings','documents', 'metadatas'])
```

```python
# SEARCH documents
vector_store.similarity_search(
    query="who among them is a bowler?",
    k=2 #k- How many of the "most" relevant documents/chunks do you want back? switch numbers and check u will understand
)
```

```python
#similarity search with score if u want to print:
vector_store.similarity_search_with_score(
    query="who among them is a bowler?",
    k=2
) #similarity score -how less it is, that doc is most relevant since it represents the distance
```

```python
# meta-data filtering
#If I only want to filter/retrieve documents based on metadata, I can leave the query empty:
vector_store.similarity_search_with_score(
    query="",
    filter={'team': 'Chennai Super Kings'} #even upper and lower cases also matter since here semantic meaning is not considered. change all to small letters and chekc the output
)
#metadata filtering is an exact/structured condition, not semantic similarity search.
# Equality filtering is case-sensitive here:
# "Chennai Super Kings" != "chennai super kings"

#"Give me documents whose metadata says team = Royal Challengers Bangalore; I don't care about semantic similarity to a query since empty."
```

```python
vector_store.similarity_search_with_score(
    query="who among them is a bowler?",
    filter={'team': 'Chennai Super Kings'}
) #nwo see the similarity scores and which came as output? but why did ms dhoni came in output he isn't a bowler cz not just query , metadata filtering is also taken into consideration.
#"First restrict the search to CSK documents, and within those documents, find the ones most similar to 'who among them is a bowler?'"
# First filter the documents to only those belonging to Chennai Super Kings, then perform semantic similarity search on those documents using the query "who among them is a bowler?" and return the most similar or top results.
```
